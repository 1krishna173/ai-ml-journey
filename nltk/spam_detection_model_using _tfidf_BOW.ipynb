To check whether the given setneces is spam or ham


```python
import nltk
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('spamhamdata.csv',sep='\t',names=['labels','messages'])
```

read the dataset then sperating using the manual columns


```python
df.head()
```

Data cleaning and preprocessing
1> lower the sentences
2> applying the stopwords
3> removing the special charecters

```python
import re
from nltk.corpus import stopwords
from nltk.stem.snowball import SnowballStemmer
from nltk.stem import WordNetLemmatizer
wnl = WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
nltk.download('stopwords')
```

```python
corpus = []
stop_words = set(stopwords.words('english'))

for i in range(0, len(df)):
    # Replace non-letters with spaces
    review = re.sub('[^a-zA-Z]', ' ', df['messages'][i])
    review = review.lower()
    review = review.split()
    # Stem words and remove stopwords
    review = [wnl.lemmatize(word,pos='v') for word in review if word not in stop_words]
    review = ' '.join(review)
    corpus.append(review)

print(f'Processed {len(corpus)} messages.')
```

```python
corpus
```

```python
###create bow (BAG OF WORDS)
from sklearn.feature_extraction.text import CountVectorizer
countv=CountVectorizer(max_features=2500,binary=True)
x=countv.fit_transform(corpus).toarray()
```

N_grams

```python
countv.vocabulary_
```

```python
y=pd.get_dummies(df['labels']).astype(int)
y=y.iloc[:,0].values
```

```python
y
```

MODEL Training

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_model_classifier=MultinomialNB().fit(x_train,y_train)
```

```python
y_pred=spam_model_classifier.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
tfidf=TfidfVectorizer(max_features=2800)
```

```python
###train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_tfidf=MultinomialNB().fit(x_train,y_train)
```

```python
y_pred=spam_tfidf.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
