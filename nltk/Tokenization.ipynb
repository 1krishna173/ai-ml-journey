```python
from nltk.tokenize import sent_tokenize,word_tokenize,TreebankWordTokenizer
```

```python
corpus='''
I am currently learning the tokenization.
I am currently a student.
My name is Krishna Mahajan.
Am I sad?
they'll be late.
'''
```

```python
sentence_token = sent_tokenize(corpus)
```

```python
sentence_token
```

```python
words=word_tokenize(corpus)
words
```

```python
tree=TreebankWordTokenizer().tokenize(corpus)
tree
```

```python
from nltk.stem import PorterStemmer
```

```python
words = ["running", "jumps", "happily", "programming"]
```

```python
porterstem=PorterStemmer()## it cuts down the prefix and suffix to the base root
```

```python
for word in words:
  print(word+'---->'+porterstem.stem(word))
```

```python
from nltk.stem import RegexpStemmer## it will work according to the regex expression
regexst=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
for word in words:
  print(word+'---->'+regexst.stem(word))
```

```python
from nltk.stem.snowball import SnowballStemmer ## choose a langauage and stem a word
snowstem=SnowballStemmer('english')
```

```python
for word in words:
  print(word+'---->'+snowstem.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer## by default the pos is noun
lemmatized=WordNetLemmatizer();
```

```python
for word in words:
  print(word+'---->'+lemmatized.lemmatize(word,pos='v'))
```
