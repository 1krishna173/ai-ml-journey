```python
## We will do the feature enginerring now (OnehotEncoding and labelEncoding)
## befire that we will divide out test and train data so that there is no data leakage
```

```python
### Dividing into independent and dependent features
x=df.drop(['total_bill'],axis=1)
y=df['total_bill']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
x_train
```

```python
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
x_train['sex']=le1.fit_transform(x_train['sex'])
x_train['smoker']=le2.fit_transform(x_train['smoker'])
x_train['time']=le3.fit_transform(x_train['time'])
```

```python
x_train
```

```python
x_test['sex']=le1.transform(x_test['sex'])
x_test['smoker']=le2.transform(x_test['smoker'])
x_test['time']=le3.transform(x_test['time'])
```

```python
x_test
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first'), [3])
    ],
    remainder='passthrough'
)
```

```python
x_train=ct.fit_transform(x_train)
```

```python
x_test=ct.transform(x_test)
```

```python
x_test
```

```python
x_train
```

                     Now we will apply the algorithm which is SVR




```python
from sklearn.svm import SVR
svr=SVR()
```

```python
svr.fit(x_train,y_train)
```

```python
y_pred=svr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,r2_score
mae=mean_absolute_error(y_test,y_pred)
print(mae)
score=r2_score(y_test,y_pred)
print(score)
```

```python
### HYPERparameter tunning using the grid search cv
from sklearn.model_selection import GridSearchCV
param_grid={
    'C':[0.1,1,10,100],
    'gamma':[1,0.1,0.001,0.00001],
    'kernel':['rbf','linear','poly']

}
```

```python
grid=GridSearchCV(svr,param_grid=param_grid,refit=True,verbose=3)
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
grid_prediction=grid.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,r2_score
mae=mean_absolute_error(y_test,grid_prediction)
print(mae)
score=r2_score(y_test,grid_prediction)
print(score)
```
