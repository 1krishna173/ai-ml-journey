```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/krishnaik06/playstore-Dataset/main/googleplaystore.csv')
```

```python
df.head()
```

```python
df_copy=df.copy()
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
df_copy.drop(df_copy.index[10472],inplace=True)
```

```python
df_copy['Reviews'].astype(int)
```

```python
df.head(3)
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size']=df_copy['Size'].str.replace('M','000')
```

```python
df_copy['Size']=df_copy['Size'].str.replace('k','')
```

```python
df_copy['Size']=df_copy['Size'].replace('Varies with device',np.nan)
```

```python
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy.head(3)
```

```python
df_copy['Installs']=df_copy['Installs'].str.replace('+','')
df_copy['Installs']=df_copy['Installs'].str.replace(',','')
```

```python
df_copy['Installs']=df_copy['Installs'].astype(int)
```

```python
df_copy['Price']=df_copy['Price'].str.replace('$','')
```

```python
df_copy['Price']=df_copy['Price'].astype(float)
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['updated_year']=df_copy['Last Updated'].dt.year
df_copy['updated_month']=df_copy['Last Updated'].dt.month
df_copy['updated_day']=df_copy['Last Updated'].dt.day
```

```python
df_copy.info()
```

```python
df_copy.to_csv('google_cleaned.csv')
```
