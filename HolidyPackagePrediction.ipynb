```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df['Gender']=df['Gender'].replace('Fe Male','Female')
```

```python
df['MaritalStatus']=df['MaritalStatus'].replace('Unmarried','Single')
```

```python
features_with_na=[column for column in df.columns if df[column].isnull().sum()>=1 ]
```

```python
features_with_na
```

```python
df.info()
```

```python
## if it is an integer or an float object we will replace it with the median
## if it is an categorical feature we will replace it with the mode
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['TypeofContact']=df['TypeofContact'].fillna(df.TypeofContact.mode()[0])
```

```python
df['DurationOfPitch'] = df['DurationOfPitch'].fillna(df['DurationOfPitch'].median())
df['NumberOfFollowups'] = df['NumberOfFollowups'].fillna(df['NumberOfFollowups'].median())
df['PreferredPropertyStar'] = df['PreferredPropertyStar'].fillna(df['PreferredPropertyStar'].median())
df['NumberOfTrips'] = df['NumberOfTrips'].fillna(0)
df['NumberOfChildrenVisiting'] = df['NumberOfChildrenVisiting'].fillna(df['NumberOfChildrenVisiting'].median())
df['MonthlyIncome'] = df['MonthlyIncome'].fillna(df['MonthlyIncome'].median())
```

```python
df.head()
```

```python
df['Visiting']=df['NumberOfPersonVisiting']+df['NumberOfChildrenVisiting']
df.drop(['NumberOfPersonVisiting','NumberOfChildrenVisiting'],axis=1,inplace=True)
```

```python
df.drop(['CustomerID'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.head()
```

```python
x=df.drop(['ProdTaken'],axis=1)
y=df['ProdTaken']
```

```python
### Train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
cat_features=x.select_dtypes(include='object').columns
num_features=x.select_dtypes(exclude='object').columns
```

```python
cat_features,num_features
```

```python
from pandas.core.arrays import categorical
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
numeric_transformer=StandardScaler()
categorical_transformer=OneHotEncoder(drop='first')
ct=ColumnTransformer(
    [
        ('StandardScaler',numeric_transformer,num_features),
        ('OneHotEncoder',categorical_transformer,cat_features)
    ]
)
```

```python
x_train=ct.fit_transform(x_train)
x_test=ct.transform(x_test)
```

```python
x_train
```

Model Traning


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report,accuracy_score,roc_auc_score
```

```python
models={
    "Random Forest": RandomForestClassifier(),
    "Decision Tree":DecisionTreeClassifier()
}
for i in range(len(models)):
  model=list(models.values())[i]
  model.fit(x_train,y_train)
  # Make Prediction
  y_train_pred=model.predict(x_train)
  y_test_pred=model.predict(x_test)
 # train performance
  model_train_accuracy=accuracy_score(y_train,y_train_pred)
  model_classification=classification_report(y_train,y_train_pred)
  model_roc=roc_auc_score(y_train,y_train_pred)
  ## test set performance
  model_test_accuracy=accuracy_score(y_test,y_test_pred)
  model_test_classification=classification_report(y_test,y_test_pred)
  model_test_roc=roc_auc_score(y_test,y_test_pred)
  print(list(models.values())[i])
  #training performace for the training dataset
  print('Model Performance for the train Dataset')
  print('accuracy',model_train_accuracy)
  print('Classification Report',model_classification)
  print('Roc_auc_score',model_roc)

  ##test performance for the trianing dataset
  print('Model Performance for the test Dataset')
  print('accuracy',model_test_accuracy)
  print('Classification Report',model_test_classification)
  print('Roc_auc_score',model_test_roc)
  print()
  print()




```
